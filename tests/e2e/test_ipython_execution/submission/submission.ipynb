```python
import matplotlib.pyplot as plt
```

```python
def generateRange(x_low, x_high, y_low, y_high):
    return list(range(x_low, x_high)), list(range(y_low, y_high))
```

```python
plt.plot(*generateRange(0,10,0,10))
plt.show()
```
